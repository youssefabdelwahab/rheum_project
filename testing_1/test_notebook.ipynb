```python
import json 
import csv 
```

```python
csv_file = '/project/def-amingh/rheum_project_data/paper_database/april_08_2026/master/paper_journal_info_2026-04-18.csv'
json_file = '/project/def-amingh/rheum_project_data/paper_database/april_08_2026/master/paper_journal_info_2026-04-18.json'

with open(csv_file, 'r') as f, open(json_file, 'w') as json_f:
    reader = csv.DictReader(f)
    for row in reader:
        json_f.write(json.dumps(row) + '\n')
```
